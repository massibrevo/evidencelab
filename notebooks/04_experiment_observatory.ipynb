# Experiment observatory

Explore the **published CPU reference experiment** with Matplotlib. These are measurements from the original workstation, not a new run on this machine. Change the parameters below, then rerun the plotting cells. No model downloads or training.

```python
import json
from pathlib import Path
from importlib.resources import files
from IPython.display import Image, display
from evidencelab.service import ROOT
from evidencelab.figures import render_benchmark

reference = ROOT / "docs/experiments/scifact-cpu.json"
if not reference.exists():
    reference = files("evidencelab") / "resources/reference.json"
run = json.loads(reference.read_text())
run["provenance"] = "reference"
metric = "recall"  # recall, mrr, ndcg
palette = "spectrum"  # spectrum, ocean, accessible
print(f"Reference experiment: {run['query_count']} queries, k={run['k']}")
```

## Scores with uncertainty

Bars compare Recall@k, MRR@k and nDCG@k. Error bars show the recorded 95% query bootstrap intervals. The reference subset is selected development data; it cannot establish generalization.

```python
display(Image(render_benchmark(run, chart="scores", palette=palette)))
```

## Query-level failure map

Rows are retrieval strategies. Columns align identical query IDs. Inspect failures individually before drawing conclusions from the averages.

```python
display(Image(render_benchmark(run, chart="heatmap", metric=metric, palette=palette)))
```

## Export a paper-ready figure

The next cell writes to the notebook workspace working directory. SVG/PDF preserve vector artwork; PNG is useful for quick reports.

```python
output = Path("reference-scores.svg")
output.write_bytes(render_benchmark(run, chart="scores", fmt="svg", theme="paper", palette="accessible"))
print(f"Saved {output.resolve()}")
```
