## Why pair observations?

Two systems answer the same queries. Query difficulty induces correlation. Resample the per-query difference, not two independent means. The percentile bootstrap estimates uncertainty in the selected query distribution; it does not remove selection bias.

```python
from evidencelab.evaluation import bootstrap
import numpy as np
# Pedagogical synthetic values only.
baseline=np.array([0,1,1,0,1,1,0,1],dtype=float)
candidate=np.array([1,1,1,0,1,1,0,1],dtype=float)
paired=bootstrap((candidate-baseline).tolist(),seed=42)
print("Synthetic paired example:",paired)
assert paired["low"] <= paired["mean"] <= paired["high"]
```

## Sample size and power

If only 24 selected development claims are evaluated, a change of one query is 4.17 percentage points. A perfect score can produce a degenerate bootstrap interval: it does not prove universal correctness. Correlated claims about one paper require clustered resampling for defensible external claims.

```python
n=24
print("One query changes mean by",100/n,"percentage points")
print("Degenerate sample:",bootstrap([1.]*n))
```
